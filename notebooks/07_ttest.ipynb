```python
import pandas as pd
from scipy import stats
```

```python
us = pd.read_csv('../data/processed/user_summary.csv')
q80 = us['WearRate'].quantile(0.8)
q20 = us['WearRate'].quantile(0.2)

top = us[us['WearRate'] >= q80]
bottom = us[us['WearRate'] <= q20]
print(f'상위 그룹 n={len(top)} (WearRate>={q80:.2f}) / 하위 그룹 n={len(bottom)} (WearRate<={q20:.2f})')
```

```python
metrics = ['AvgSteps','AvgCalories','AvgActiveMinutes','AvgSedentaryRatio','AvgVeryActiveMinutes','AvgSleepMinutes']
rows = []
for m in metrics:
    t_top = top[m].dropna(); t_bot = bottom[m].dropna()
    if len(t_top) < 2 or len(t_bot) < 2:
        continue
    t, p = stats.ttest_ind(t_top, t_bot, equal_var=False)
    rows.append([m, t_top.mean(), t_bot.mean(), t, p, 'Y' if p < 0.05 else ''])
```

```python
result = pd.DataFrame(rows, columns=['지표','상위평균','하위평균','t','p','유의(p<.05)'])
result
```
